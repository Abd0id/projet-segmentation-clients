```python
import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
from scipy.stats import skew
```

```python
df = pd.read_csv('../data/raw/dataset-6ab989b678d90892980305.csv')
```

```python
numeric_cols = ["BALANCE", "PURCHASES", "ONEOFF_PURCHASES", "INSTALLMENTS_PURCHASES",
                 "CASH_ADVANCE", "CREDIT_LIMIT", "PAYMENTS"]

fig, axes = plt.subplots(4, 2, figsize=(14, 16))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.histplot(df[col].dropna(), kde=True, ax=axes[i])
    col_skew = skew(df[col].dropna())
    axes[i].set_title(f"{col} (skew = {col_skew:.2f})")

axes[7].axis("off")

plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(10, 8))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix - Credit Card Customer Data")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 2, figsize=(14, 16))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.boxplot(x=df[col], ax=axes[i])
    axes[i].set_title(col)

axes[7].axis("off")
plt.tight_layout()
plt.show()
```
